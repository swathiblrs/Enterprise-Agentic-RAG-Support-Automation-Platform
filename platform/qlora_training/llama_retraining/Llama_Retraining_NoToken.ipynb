**Prepared, not trained.** Same public Llama-3.2-1B-Instruct checkpoint, no token or Secrets setup. Uses 4-bit QLoRA with six-class label-token cross-entropy, label smoothing and development macro-F1 checkpoint selection.

The same 612 training and 210 development cases are retained. Untuned Llama, the old adapter and the new adapter are evaluated with the same single-pass scorer on the previously inspected synthetic benchmark. No new blind-test or production claim is made.

## Upload the retraining bundle
Upload **llama_retraining_bundle.zip**. It includes synthetic data, code and your already-downloaded prior adapter for a same-session comparison. It contains no base-model weights or credentials.


```python
from google.colab import files
import io
import stat
import zipfile
uploaded = files.upload()
if len(uploaded) != 1:
    raise ValueError('Upload only llama_retraining_bundle.zip')
name, content = next(iter(uploaded.items()))
if not name.endswith('.zip'):
    raise ValueError('Expected a ZIP bundle')
root = Path('/content/llama_retraining')
if root.exists():
    raise RuntimeError('Folder exists. Use a fresh runtime to avoid mixing runs.')
with zipfile.ZipFile(io.BytesIO(content)) as archive:
    if sum(i.file_size for i in archive.infolist()) > 64*1024**2:
        raise ValueError('Unexpected archive size; stop rather than extracting an unrecognized bundle')
    for member in archive.infolist():
        target = (root/member.filename).resolve()
        if root.resolve() not in target.parents or stat.S_ISLNK(member.external_attr >> 16):
            raise ValueError('Unsafe archive entry')
    archive.extractall(root)
sys.path.insert(0, str(root))
sys.modules.pop('protocol', None)
from protocol import verify_bundle
manifest = verify_bundle(root)
print(manifest['provenance'])
```

## Check public model access: no token
The public checkpoint is checked anonymously before weight downloads. If the repository becomes gated/private or is unavailable, stop; there is no login workaround or paid fallback. The Meta Llama license still applies.


```python
import json
from huggingface_hub import HfApi, hf_hub_download
repo_id = 'unsloth/Llama-3.2-1B-Instruct'
revision = '5a8abab4a5d6f164389b1079fb721cfab8d7126c'
try:
    info = HfApi().model_info(repo_id, revision=revision, token=False)
    if info.private or info.gated:
        raise RuntimeError('This checkpoint is no longer anonymously available')
    config_path = hf_hub_download(repo_id=repo_id, filename='config.json', revision=revision, token=False)
    if json.loads(Path(config_path).read_text()).get('model_type') != 'llama':
        raise RuntimeError('The public checkpoint is not a standard Llama model')
except Exception:
    raise RuntimeError('Anonymous model check failed. Stop here; share the error, not any credentials.') from None
print('Public Llama checkpoint verified. No token was read or used.')
```

## Retrain and compare fairly
Up to 300 steps; learning rate 0.00005; rank16; label smoothing0.05. Best checkpoint is selected using development macro-F1. Inputs stop before the class token, so target labels cannot leak into the prediction.

Both old and new adapters use the same six-token scorer in this session. Differences from the old six-pass JSON scoring are not training-only gains. Qwen and lightweight-model metrics remain historical references. The benchmark is not a new blind test, and no model is promoted automatically.


```python
results = Path('/content/llama_retrained_results')
subprocess.run([sys.executable, str(root/'retrain.py'), 'run', '--out', str(results)], check=True)
print((results/'RESULTS.md').read_text())
```
